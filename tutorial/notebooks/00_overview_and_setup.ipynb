Use this helper in later notebooks when you want to inspect a command before deciding whether to execute it.


```python
import shlex
import subprocess

def show_command(parts):
    print(' '.join(shlex.quote(str(p)) for p in parts))

def maybe_run(parts, *, enabled=False, cwd=REPO_ROOT):
    show_command(parts)
    if not enabled:
        print('skipped: enable the guard flag before running live commands')
        return None
    return subprocess.run(parts, cwd=cwd, text=True, check=True, capture_output=True)
```
